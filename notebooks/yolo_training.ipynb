```python
# YOLOv8 多类别训练（银边 + 异物）

## 1. 环境检查与依赖
import torch
import yaml
from pathlib import Path
from ultralytics import YOLO
#import matplotlib.pyplot as plt
#import cv2
import os

os.chdir('..')   # 切换到项目根目录
print("当前工作目录:", os.getcwd())

print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU device: {torch.cuda.get_device_name()}")
```

```python
## 2. 数据集配置检查
data_yaml_path = Path("./datasets/visible/data.yaml")
if not data_yaml_path.exists():
    raise FileNotFoundError(f"未找到 {data_yaml_path}，请检查路径")

with open(data_yaml_path) as f:
    data_cfg = yaml.safe_load(f)

print("数据集配置:")
for k, v in data_cfg.items():
    print(f"  {k}: {v}")

# 验证目录存在
for split in ["train", "val"]:
    img_dir = Path(data_cfg.get('path', '.')) / data_cfg[split]
    lbl_dir = img_dir.parent.parent / "labels" / split
    print(f"{split} 图片数量: {len(list(img_dir.glob('*.*')))}")
    print(f"{split} 标签数量: {len(list(lbl_dir.glob('*.txt')))}")
```

```python
## 3. 加载预训练模型
# 加载配置文件
cfg_path = Path("./configs/yolo_config.yaml")
with open(cfg_path, 'r') as f:
    train_config = yaml.safe_load(f)

# 查看配置
print("加载的训练参数：")
for k, v in train_config.items():
    print(f"{k}: {v}")

# 加载模型并训练
model = YOLO('yolov8n.pt')
results = model.train(**train_config)
```
